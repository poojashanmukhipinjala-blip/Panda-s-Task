Task 3 — Dynamic Sales Classification
Dataset: E-Commerce Orders

Concepts: Conditional transformation, np.where(), np.select()
1)  Create Sales_Level: Sales < 5000 = Low; 5000–20000 = Medium; 20000–50000 = High; Sales > 50000 = Very High.
2)  Create Customer_Type: Rating >= 4 = Satisfied; Rating < 4 = Needs Attention.
3)  Display Order_ID, Sales, Rating, Sales_Level, and Customer_Type.

```python
def Sales_Level(sales):
  if sales < 5000:
    return 'Low'
  elif sales < 20000:
    return 'Medium'
  elif sales < 50000:
    return 'High'
  else:
    return 'Very High'
```

```python
orders['Sales_level'] = orders['sales'].apply(Sales_Level)
```

```python
orders.head(1)
```

```python
def customer_rating (Rating):
  if Rating >= 4:
    return "Satisfied"
  else:
    return 'Needs Attention'
```

```python
orders['Customer_Ratings'] = orders['Rating'].apply(customer_rating)
```

Task 10 — Complete Analyst Challenge
Dataset: E-Commerce Orders
Concepts: Combined Day 1–8 workflow; no groupby() or aggregation

1) Create Revenue = Quantity × Unit_Price.
2) Create Discount_Rate: Revenue >= 50000 = 15%; Revenue >= 20000 = 10%; otherwise 5%.
3) Calculate Discount_Amount and Final_Revenue.
4) Create Customer_Status: Rating >= 4 = Satisfied; otherwise Needs Attention.
5) Find orders where Category = Electronics AND Final_Revenue > 20000 AND Rating >= 4 AND Payment_Mode = UPI.
6) Sort qualifying orders by Final_Revenue descending.
7) Return Order_ID, Customer, City, Product, Quantity, Revenue, Discount_Amount, Final_Revenue, and Rating.
8) Create Order_Label such as: ORD001 - Laptop - ₹55250.

```python
orders['Revenue'] = orders['Quantity'] * orders['Unit_Price']
```

```python
conditions = [
    orders['Revenue'] >= 50000,
    orders['Revenue'] >= 20000
]
rates = [0.15, 0.10]

orders['Discount_Rate'] = np.select(conditions, rates, default=0.05)

orders['Discount_Amount'] = orders['Revenue'] * orders['Discount_Rate']
orders['Final_Revenue'] = orders['Revenue'] - orders['Discount_Amount']

```

```python
def get_customer_ratings(Rating):
  if Rating >= 4:
    return 'Satisfied'
  else:
    return'Attention'
```

```python
orders['Customer_Rating'] = orders['Rating'].apply(get_customer_ratings)
```

```python
orders[(orders['Category'] == 'Electonics') & (orders['Final_Revenue']>20000) & (orders['Rating'] >= 4) & (orders['Payment_Mode'] == 'UPI')]
```

```python
orders.loc[:,['Order_ID', 'Customer', 'City', 'Product', 'Quantity', 'Revenue', 'Discount_Amount', 'Final_Revenue','Rating']]
```
